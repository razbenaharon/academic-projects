## **Q1**

```python
from typing_extensions import clear_overloads
from scipy.stats import norm
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# from project A
df = pd.read_csv('hotel_bookings.csv')
selected_columns = ['hotel', 'lead_time', 'arrival_date_month', 'adults', 'children', 'babies', 'is_canceled', 'adr', 'total_of_special_requests', 'stays_in_week_nights', 'stays_in_weekend_nights']
df_selected = df[selected_columns]
df_sampled = df_selected.sample(n=10000, random_state=7)


print('A.')
lead_time_canceled_full_df = df_sampled[df_sampled['is_canceled'] == 1]['lead_time']
lead_time_not_canceled_full_df=df_sampled[df_sampled['is_canceled'] == 0]['lead_time']
print('Catagories Averages:')
print(lead_time_canceled_full_df.mean())
print(lead_time_not_canceled_full_df.mean())

print('')
print('B.')

z_critical = norm.ppf(0.975)  # Critical z-value for 95% confidence

n= len(lead_time_canceled_full_df)
mean_canceled = lead_time_canceled_full_df.mean()
std_canceled =lead_time_canceled_full_df.std()
margin_of_error_canceled = z_critical * (std_canceled / np.sqrt(n))
confidence_interval_canceled = (mean_canceled - margin_of_error_canceled, mean_canceled + margin_of_error_canceled)
print (confidence_interval_canceled)

m = len(lead_time_not_canceled_full_df)
mean_not_canceled = lead_time_not_canceled_full_df.mean()
std_not_canceled =lead_time_not_canceled_full_df.std()
margin_of_error_not_canceled = z_critical * (std_canceled / np.sqrt(m))
confidence_interval_not_canceled = (mean_not_canceled - margin_of_error_not_canceled, mean_not_canceled + margin_of_error_not_canceled)
print (confidence_interval_not_canceled)


```

```python
from scipy.stats import f
from scipy.stats import t
from scipy.stats import chi2

import math

S_canceled_squered = std_canceled**2
S_not_canceled_squered = std_not_canceled**2

# B
if S_canceled_squered>S_not_canceled_squered:
  F=S_canceled_squered/S_not_canceled_squered
  f_value = f.ppf(0.95, n-1, m-1)
else:
  F=S_not_canceled_squered/S_canceled_squered
  f_value = f.ppf(0.95, m-1, n-1)
print (F>f_value)


# C
S_p_squered= ((n-1)*S_canceled_squered+(m-1)*S_not_canceled_squered)/(n+m-2)
T= (mean_canceled-mean_not_canceled)/(np.sqrt(S_p_squered)*np.sqrt((1/n)+(1/m)))
t_value = t.ppf(0.95,n+m-2)
print (T>t_value)

# D
W= abs((mean_canceled-mean_not_canceled)/np.sqrt((S_canceled_squered/n)+(S_not_canceled_squered/m)))
z = norm.ppf(0.975)
print (W>z)

#E
p_val_t = 1-t.cdf(T,n+m-2)
p_val_z = 2*min(norm.cdf(W),1-norm.cdf(W))

print (p_val_t)
print (p_val_z)

#F
sample_variance=df_sampled['lead_time'].var()
mean_all=df_sampled['lead_time'].mean()
first = (n+m)*math.log(sample_variance/S_p_squered)
second= 0
for i in range(n):
  second += (lead_time_canceled_full_df.values[i]-mean_all)**2/sample_variance - (lead_time_canceled_full_df.values[i]-mean_canceled)**2/S_p_squered
third=0
for i in range(m):
  third+= (((lead_time_not_canceled_full_df.values[i]-mean_all)**2/sample_variance) - ((lead_time_not_canceled_full_df.values[i]-mean_not_canceled)**2/S_p_squered))

statistic = first + second + third
test_quantile = chi2.ppf(0.95,1)
print(statistic > test_quantile)



```

```python
#  D
import matplotlib.pyplot as plt

avg_p_val=[]
avg_canceled_margin=[]
avg_not_canceled_margin=[]

canceled_avg_percentages={}
not_canceled_avg_percentages={}


for i in [30,50,100,500]:
    canceled_margin_list=[]
    not_canceled_margin_list=[]
    p_val_list=[]
    canceled_avg_percentages[i]=0
    not_canceled_avg_percentages[i]=0


    for j in range (100):
      df_sampled=df_selected.sample(n=i)
      lead_time_canceled_df = df_sampled[df_sampled['is_canceled'] == 1]['lead_time']
      lead_time_not_canceled_df=df_sampled[df_sampled['is_canceled'] == 0]['lead_time']

      n = len(lead_time_canceled_df)
      mean_canceled = lead_time_canceled_df.mean()
      std_canceled =lead_time_canceled_df.std()
      margin_of_error_canceled = z_critical * (std_canceled / np.sqrt(n))
      canceled_margin_list.append(2*margin_of_error_canceled)
      confidence_interval_canceled = (mean_canceled - margin_of_error_canceled, mean_canceled + margin_of_error_canceled)
      if confidence_interval_canceled[0] <= lead_time_canceled_full_df.mean() <= confidence_interval_canceled[1]:
        canceled_avg_percentages[i]+=1


      m = len(lead_time_not_canceled_df)
      mean_not_canceled = lead_time_not_canceled_df.mean()
      std_not_canceled =lead_time_not_canceled_df.std()
      margin_of_error_not_canceled = z_critical * (std_canceled / np.sqrt(m))
      not_canceled_margin_list.append(2*margin_of_error_not_canceled)
      confidence_interval_not_canceled = (mean_not_canceled - margin_of_error_not_canceled, mean_not_canceled + margin_of_error_not_canceled)
      if confidence_interval_not_canceled[0] <= lead_time_not_canceled_full_df.mean() <= confidence_interval_not_canceled[1]:
        not_canceled_avg_percentages[i]+=1


      W= abs((mean_canceled-mean_not_canceled)/np.sqrt((S_canceled_squered/n)+(S_not_canceled_squered/m)))
      p_val_z = 2*min(norm.cdf(W),1-norm.cdf(W))
      p_val_list.append(p_val_z)


    avg_canceled_margin.append(np.mean(canceled_margin_list))
    avg_not_canceled_margin.append(np.mean(not_canceled_margin_list))
    avg_p_val.append(np.mean(p_val_list))


print ('canceled_avg_percentages')
print(canceled_avg_percentages)
print ('not_canceled_avg_percentages')
print (not_canceled_avg_percentages)

print ('')

sample_sizes = [30, 50, 100, 500]
fig, axes = plt.subplots(3, 1, figsize=(10, 18))

# Graph 1: avg_p_val
axes[0].plot(sample_sizes, avg_p_val, color='r', marker='o', linestyle='-', linewidth=2, markersize=8)
axes[0].set_title('P-Value at Different Sample Sizes')
axes[0].set_xlabel('Sample Size')
axes[0].set_ylabel('P-Value')
axes[0].set_xticks(sample_sizes)  # Set x-ticks at the positions of the points
axes[0].grid(True)

# Graph 2: avg_canceled_margin
axes[1].plot(sample_sizes, avg_canceled_margin, color='b', marker='o', linestyle='-', linewidth=2, markersize=8)
axes[1].set_title('Canceled Margin at Different Sample Sizes')
axes[1].set_xlabel('Sample Size')
axes[1].set_ylabel('Canceled Margin')
axes[1].set_xticks(sample_sizes)  # Set x-ticks at the positions of the points
axes[1].grid(True)

# Graph 3: avg_not_canceled_margin
axes[2].plot(sample_sizes, avg_not_canceled_margin, color='g', marker='o', linestyle='-', linewidth=2, markersize=8)
axes[2].set_title('Not Canceled Margin at Different Sample Sizes')
axes[2].set_xlabel('Sample Size')
axes[2].set_ylabel('Not Canceled Margin')
axes[2].set_xticks(sample_sizes)  # Set x-ticks at the positions of the points
axes[2].grid(True)

plt.tight_layout()
plt.show()
```
